# MoE LLM — Mixture-of-Experts Language Model from Scratch

A from-scratch, end-to-end implementation of a decoder-only **Mixture-of-Experts (MoE)**
transformer in PyTorch: data pipeline, model, training loop, and text generation.

This is an original implementation of the general MoE-LLM recipe (GQA + top-k
routing + load-balancing loss + weight tying, trained on WikiText with a BPE
tokenizer) — the same recipe used by models like Mixtral and DeepSeekMoE —
rather than a copy of any single tutorial's code.

## Architecture

- **Three swappable attention variants**, all with RoPE positional encoding and causal masking:
  - **MHA** — standard multi-head attention (`attention.py`, `num_kv_heads == num_heads`)
  - **GQA** — Grouped Query Attention (`attention.py`) — query heads split into groups sharing one uncompressed K/V
  - **GLA** — Grouped Latent Attention (`grouped_latent_attention.py`) — query heads split into groups, each caching a compressed latent instead of full K/V
- **Mixture-of-Experts feed-forward layer** (`moe.py`):
  - `Expert`: a SwiGLU feed-forward network (Llama-style, no bias terms)
  - `Router`: scores every token against all experts, picks the top-k, and
    renormalizes their weights
  - `MoEFeedForward`: dispatches tokens to their chosen experts, combines
    weighted outputs, and computes a **Switch-Transformer-style load-balancing
    auxiliary loss** so training doesn't collapse onto a handful of experts
- **MoELLM** (`model.py`): stacks pre-RMSNorm blocks of (attention + FFN), with
  **weight tying** between the token embedding and the output projection, and
  no bias terms anywhere (Llama-style).
- **Data pipeline** (`data.py`): downloads WikiText-103 from Hugging Face,
  strips blank/heading lines and `<unk>` markers, joins articles with an
  `<|endoftext|>` separator, and tokenizes with GPT-2's BPE tokenizer
  (`tiktoken`). Falls back to a tiny bundled text sample if the dataset/tokenizer
  can't be downloaded (e.g. no internet access), so the pipeline can still be
  smoke-tested offline.
- **Training** (`train.py`): AdamW + cosine LR schedule, gradient clipping,
  cross-entropy loss plus a weighted load-balancing auxiliary loss, periodic
  validation, checkpointing.
- **Generation** (`generate.py`): autoregressive sampling with temperature and
  top-k filtering, loaded from a saved checkpoint.

## Evaluate

```bash
python eval.py --checkpoint checkpoint.pt --split test
```

This reports:
- **Language modeling quality**: cross-entropy loss, perplexity, bits-per-token,
  and next-token top-1 accuracy on a held-out split (`test` by default; also
  supports `validation`/`train`).
- **Per-layer expert load-balance**: what fraction of tokens each expert in
  each MoE layer actually received, plus a `balance_score` (1.0 = perfectly
  uniform usage across experts, up to `num_experts` if everything collapsed
  onto a single expert). This is the standard way to check a MoE model isn't
  quietly wasting most of its experts.
- **Qualitative samples**: a few greedy/sampled completions from fixed prompts
  for a manual sanity check alongside the numbers.

Useful flags:
- `--max_batches N` — evaluate on only the first N batches, for a quick check
  instead of the full split
- `--output_json report.json` — dump the full metrics + per-layer report +
  samples as JSON (handy for comparing checkpoints/runs over time)

## Comparing architectures: MHA vs GQA vs GLA, MoE vs MLP

The model is config-driven (`attention_type`: `mha`/`gqa`/`gla`, `ffn_type`: `moe`/`dense`
— `mlp` is accepted as a synonym for `dense`), so you can train any of the
3 × 2 = 6 combinations and compare them fairly.

**What each attention option is:**
- **`mha`** — standard Multi-Head Attention: every head has its own K/V. No cache savings; the quality baseline.
- **`gqa`** — Grouped Query Attention: heads split into groups, each group shares one *uncompressed* K/V. Trivial to shard (one group per device).
- **`gla`** — **Grouped Latent Attention** (Zadouri, Strauss & Dao 2025, [arxiv.org/abs/2505.21487](https://arxiv.org/abs/2505.21487)) — heads split into groups like GQA, but each group caches a *compressed latent* instead of full K/V, reconstructing K/V from it on demand. Positioned between GQA (no compression, easy to shard) and MLA (max compression, one global latent, harder to shard) — GLA aims for MLA-level compression with GQA-level shardability.
  - ⚠️ Naming collision: the paper also abbreviates this "GLA," which is unrelated to *Gated Linear Attention* (Yang et al. 2023, a linear-attention variant) — that older module is still in this repo as `gated_linear_attention.py` but is **not** part of the active comparison.

**Train one config:**
```bash
python train.py --attention mha --ffn dense    # baseline
python train.py --attention gqa --ffn moe
python train.py --attention gla --ffn moe      # Grouped Latent Attention + MoE
```

**Train and compare all 6 in one run** (recommended — guarantees every
config sees identical data batches and the same random seed):
```bash
python compare.py --max_steps 2000 --save_checkpoints
```
This prints a table of total params, active params/token, **KV-cache elements
per token**, val loss/perplexity, and throughput — and writes full results to
`comparison.json`. Use `--configs mha:dense gla:moe` to run only a subset.

**Then evaluate any saved checkpoint in depth** with `eval.py` as usual — it
reports the same KV-cache/params/throughput numbers plus perplexity, accuracy,
and (for MoE models) the per-layer expert-balance report.

### Why report KV-cache size at all
This is the actual point of MHA vs GQA vs GLA — they're not really competing
on raw quality (all three can be made to match with enough tuning), they're
competing on **how much has to be cached per generated token**, since that's
what limits batch size and context length in real serving. Expect roughly:
`kv_cache(mha) > kv_cache(gqa) > kv_cache(gla)` — if GLA doesn't come out
smallest in your run, something's misconfigured (check `NUM_LATENT_GROUPS`/
`LATENT_DIM` vs `NUM_KV_HEADS` in `train.py`).

### Why `--max_steps`, not just `--epochs`
Comparing "1 epoch of GQA" to "1 epoch of GLA" isn't a fair fight if one is
faster per step — you'd really be comparing compute budgets. Pick a fixed
`--max_steps` (or a fixed wall-clock budget) so every config gets the same
amount of training, and report throughput as a separate axis rather than
mixing it into the training budget.

### Why the dense/MLP FFN width is `top_k * expert_hidden_dim`
This is what "iso-compute" means in `ffn.py`'s
`dense_hidden_dim_matching_active_compute`: a token that visits `top_k`
experts of width `H` each does exactly as much FFN computation as a dense
layer of width `top_k * H`. Comparing MoE against a dense layer of the
*same total width as all experts combined* would be comparing more compute
against less — the interesting MoE result is "does routing to a subset of a
much larger total capacity beat a same-compute dense layer," which is what
`total_params` vs `active_params` in the comparison table shows directly.

### Caveat on this GLA implementation
This is a correctness-focused reference implementation of Grouped Latent
Attention's core idea (per-group down/up projection through a compressed
latent), not the paper's production kernel. In particular it recomputes K/V
from the latent on every forward pass rather than implementing a real
incremental-decoding cache — so `kv_cache_elements_per_token()` reports the
theoretical cache size correctly, but you won't see the paper's reported
end-to-end serving speedups from this module directly; those come from a
custom kernel doing the caching, which is out of scope here.

## Notes on scaling up

- Total params grow with `num_experts`, but the compute per token only grows
  with `top_k` — that's the entire point of MoE. Try more/larger experts
  before making the rest of the model bigger.
- `AUX_LOSS_WEIGHT` in `train.py` controls how strongly the router is pushed
  toward balanced expert usage. Too low → a few experts absorb most tokens
  (wasted capacity); too high → hurts task performance. `0.01` is a common
  starting point.
- `NUM_KV_HEADS` controls the GQA group size (`num_heads / num_kv_heads`).
  Setting it equal to `num_heads` recovers standard multi-head attention;
  setting it to `1` recovers multi-query attention.

````python
%%writefile attention.py

"""
Attention building blocks for the MoE LLM:
  - RMSNorm (used instead of LayerNorm, no bias, Llama-style)
  - RotaryEmbedding (RoPE) for relative positional information
  - Causal Grouped-Query Attention (GQA)

GQA reduces KV-cache / parameter cost vs. full multi-head attention by sharing
each Key/Value head across a *group* of Query heads. num_heads must be
divisible by num_kv_heads.
"""
import math
import torch
import torch.nn as nn


class RMSNorm(nn.Module):
    """Root-Mean-Square LayerNorm: normalizes by RMS instead of mean/var,
    and has no bias term (Llama-style)."""

    def __init__(self, dim: int, eps: float = 1e-6):
        super().__init__()
        self.eps = eps
        self.weight = nn.Parameter(torch.ones(dim))

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        rms = x.pow(2).mean(dim=-1, keepdim=True).add(self.eps).rsqrt()
        return x * rms * self.weight


def build_rope_cache(head_dim: int, max_seq_len: int, base: float = 10000.0):
    """Precompute cos/sin tables for rotary position embeddings."""
    inv_freq = 1.0 / (base ** (torch.arange(0, head_dim, 2).float() / head_dim))
    t = torch.arange(max_seq_len).float()
    freqs = torch.outer(t, inv_freq)  # [max_seq_len, head_dim/2]
    emb = torch.cat([freqs, freqs], dim=-1)  # [max_seq_len, head_dim]
    return emb.cos(), emb.sin()


def rotate_half(x: torch.Tensor) -> torch.Tensor:
    x1, x2 = x.chunk(2, dim=-1)
    return torch.cat([-x2, x1], dim=-1)


def apply_rope(x: torch.Tensor, cos: torch.Tensor, sin: torch.Tensor) -> torch.Tensor:
    """x: [batch, heads, seq_len, head_dim]; cos/sin: [seq_len, head_dim]"""
    cos = cos[None, None, :, :]
    sin = sin[None, None, :, :]
    return x * cos + rotate_half(x) * sin


class GroupedQueryAttention(nn.Module):
    """Causal self-attention where K/V heads are shared across groups of
    Q heads, with RoPE positional encoding and no linear-layer biases."""

    def __init__(self, embed_dim: int, num_heads: int, num_kv_heads: int,
                 max_seq_len: int, dropout: float = 0.0):
        super().__init__()
        assert embed_dim % num_heads == 0, "embed_dim must divide evenly by num_heads"
        assert num_heads % num_kv_heads == 0, "num_heads must divide evenly by num_kv_heads"

        self.num_heads = num_heads
        self.num_kv_heads = num_kv_heads
        self.head_dim = embed_dim // num_heads
        self.group_size = num_heads // num_kv_heads

        self.q_proj = nn.Linear(embed_dim, num_heads * self.head_dim, bias=False)
        self.k_proj = nn.Linear(embed_dim, num_kv_heads * self.head_dim, bias=False)
        self.v_proj = nn.Linear(embed_dim, num_kv_heads * self.head_dim, bias=False)
        self.out_proj = nn.Linear(embed_dim, embed_dim, bias=False)
        self.attn_dropout = nn.Dropout(dropout)

        cos, sin = build_rope_cache(self.head_dim, max_seq_len)
        self.register_buffer("rope_cos", cos, persistent=False)
        self.register_buffer("rope_sin", sin, persistent=False)

        causal = torch.tril(torch.ones(max_seq_len, max_seq_len, dtype=torch.bool))
        self.register_buffer("causal_mask", causal, persistent=False)

    def kv_cache_elements_per_token(self) -> int:
        """How many scalars must be cached per token during generation.
        MHA is the special case num_kv_heads == num_heads."""
        return 2 * self.num_kv_heads * self.head_dim  # K and V, per KV head

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        B, T, C = x.shape

        q = self.q_proj(x).view(B, T, self.num_heads, self.head_dim).transpose(1, 2)
        k = self.k_proj(x).view(B, T, self.num_kv_heads, self.head_dim).transpose(1, 2)
        v = self.v_proj(x).view(B, T, self.num_kv_heads, self.head_dim).transpose(1, 2)

        cos, sin = self.rope_cos[:T], self.rope_sin[:T]
        q = apply_rope(q, cos, sin)
        k = apply_rope(k, cos, sin)

        # Broadcast each KV head across its group of Q heads
        k = k.repeat_interleave(self.group_size, dim=1)
        v = v.repeat_interleave(self.group_size, dim=1)

        attn_scores = (q @ k.transpose(-2, -1)) / math.sqrt(self.head_dim)
        mask = self.causal_mask[:T, :T]
        attn_scores = attn_scores.masked_fill(~mask, float("-inf"))
        attn_weights = torch.softmax(attn_scores, dim=-1)
        attn_weights = self.attn_dropout(attn_weights)

        out = attn_weights @ v
        out = out.transpose(1, 2).contiguous().view(B, T, C)
        return self.out_proj(out)
````

````python
%%writefile grouped_latent_attention.py


"""
Grouped Latent Attention (GLA) -- Zadouri, Strauss & Dao,
"Hardware-Efficient Attention for Fast Decoding" (2025)
https://arxiv.org/abs/2505.21487

NOTE ON NAMING: this paper also abbreviates its method "GLA", which
collides with the unrelated "Gated Linear Attention" (Yang et al., 2023,
kept in this repo as gated_linear_attention.py but not used in the default
comparison). Same three letters, different technique -- don't confuse them.

GLA sits between two existing ways of shrinking the KV cache:

  - MLA (DeepSeek-V2/V3): ALL query heads share ONE compressed latent
    vector, from which K and V are reconstructed. Maximizes compression,
    but a single latent shared by every head is awkward to shard across
    devices -- every shard needs the whole thing.
  - GQA: query heads are split into groups, each group gets its OWN
    full-size (uncompressed) K/V. Trivial to shard (one group per device),
    but no compression beyond head-sharing.

GLA groups heads like GQA, but gives each group a compressed LATENT vector
(like MLA) instead of a full K/V. Each group is self-contained -- its
down/up projections only ever touch its own latent -- which is what keeps
GLA as shardable as GQA while still compressing the cache like MLA.

Per group g, at each timestep:
    latent_t^(g) = W_down^(g) x_t                     (embed_dim -> latent_dim)
    K_t^(g)      = W_up_k^(g) latent_t^(g)             (latent_dim -> head_dim)
    V_t^(g)      = W_up_v^(g) latent_t^(g)             (latent_dim -> head_dim)

Only latent_t^(g) needs to be cached during incremental generation (K/V are
reconstructed from it on demand) -- that's the actual source of the memory
saving. This module recomputes K/V from the latent on every forward pass
for clarity, rather than implementing a custom incremental-decoding cache;
see `kv_cache_elements_per_token()` for the memory-saving accounting that
matters in practice.
"""
import math
import torch
import torch.nn as nn

from attention import build_rope_cache, apply_rope


class GroupedLatentAttention(nn.Module):
    def __init__(self, embed_dim: int, num_heads: int, num_groups: int, latent_dim: int,
                 max_seq_len: int, dropout: float = 0.0):
        super().__init__()
        assert embed_dim % num_heads == 0, "embed_dim must divide evenly by num_heads"
        assert num_heads % num_groups == 0, "num_heads must divide evenly by num_groups"

        self.num_heads = num_heads
        self.num_groups = num_groups
        self.head_dim = embed_dim // num_heads
        self.group_size = num_heads // num_groups
        self.latent_dim = latent_dim

        self.q_proj = nn.Linear(embed_dim, num_heads * self.head_dim, bias=False)

        # Independent per-group down/up projections -- each group only ever
        # reads its OWN latent, which is what makes GLA shard as cleanly as
        # GQA (one group per device) rather than needing MLA's single
        # globally-shared latent.
        self.latent_down = nn.ModuleList(
            [nn.Linear(embed_dim, latent_dim, bias=False) for _ in range(num_groups)]
        )
        self.k_up = nn.ModuleList(
            [nn.Linear(latent_dim, self.head_dim, bias=False) for _ in range(num_groups)]
        )
        self.v_up = nn.ModuleList(
            [nn.Linear(latent_dim, self.head_dim, bias=False) for _ in range(num_groups)]
        )

        self.out_proj = nn.Linear(embed_dim, embed_dim, bias=False)
        self.attn_dropout = nn.Dropout(dropout)

        cos, sin = build_rope_cache(self.head_dim, max_seq_len)
        self.register_buffer("rope_cos", cos, persistent=False)
        self.register_buffer("rope_sin", sin, persistent=False)

        causal = torch.tril(torch.ones(max_seq_len, max_seq_len, dtype=torch.bool))
        self.register_buffer("causal_mask", causal, persistent=False)

    def kv_cache_elements_per_token(self) -> int:
        """Only the latent needs caching (K/V are reconstructed from it),
        so this is num_groups * latent_dim -- not 2 * num_groups * head_dim
        as it would be for GQA's uncompressed per-group K/V."""
        return self.num_groups * self.latent_dim

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        B, T, C = x.shape

        q = self.q_proj(x).view(B, T, self.num_heads, self.head_dim).transpose(1, 2)  # [B,H,T,D]

        k_groups, v_groups = [], []
        for g in range(self.num_groups):
            latent_g = self.latent_down[g](x)   # [B, T, latent_dim]
            k_groups.append(self.k_up[g](latent_g))  # [B, T, head_dim]
            v_groups.append(self.v_up[g](latent_g))  # [B, T, head_dim]

        k = torch.stack(k_groups, dim=1)  # [B, G, T, D]
        v = torch.stack(v_groups, dim=1)  # [B, G, T, D]

        cos, sin = self.rope_cos[:T], self.rope_sin[:T]
        q = apply_rope(q, cos, sin)
        k = apply_rope(k, cos, sin)

        # Broadcast each group's K/V across its group_size query heads
        k = k.repeat_interleave(self.group_size, dim=1)  # [B, H, T, D]
        v = v.repeat_interleave(self.group_size, dim=1)

        attn_scores = (q @ k.transpose(-2, -1)) / math.sqrt(self.head_dim)
        mask = self.causal_mask[:T, :T]
        attn_scores = attn_scores.masked_fill(~mask, float("-inf"))
        attn_weights = torch.softmax(attn_scores, dim=-1)
        attn_weights = self.attn_dropout(attn_weights)

        out = attn_weights @ v
        out = out.transpose(1, 2).contiguous().view(B, T, C)
        return self.out_proj(out)
````

````python
%%writefile moe.py

"""
Mixture-of-Experts feed-forward layer.

  - Expert:  a small SwiGLU FFN (same shape every expert has, different weights)
  - Router:  a linear layer that scores every expert per-token, picks the
             top-k, and softmax-normalizes just those k scores
  - MoEFeedForward: dispatches each token to its chosen experts, combines
    their outputs weighted by router probability, and returns an auxiliary
    load-balancing loss so training doesn't collapse onto a few experts.
"""
import torch
import torch.nn as nn
import torch.nn.functional as F


class Expert(nn.Module):
    """A single feed-forward expert using a SwiGLU activation (Llama-style),
    with no bias terms."""

    def __init__(self, embed_dim: int, hidden_dim: int):
        super().__init__()
        self.gate_proj = nn.Linear(embed_dim, hidden_dim, bias=False)
        self.up_proj = nn.Linear(embed_dim, hidden_dim, bias=False)
        self.down_proj = nn.Linear(hidden_dim, embed_dim, bias=False)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.down_proj(F.silu(self.gate_proj(x)) * self.up_proj(x))


class Router(nn.Module):
    """Scores each token against every expert and selects the top-k."""

    def __init__(self, embed_dim: int, num_experts: int, top_k: int):
        super().__init__()
        self.num_experts = num_experts
        self.top_k = top_k
        self.gate = nn.Linear(embed_dim, num_experts, bias=False)

    def forward(self, x_flat: torch.Tensor):
        """
        x_flat: [num_tokens, embed_dim]
        Returns:
          topk_weights: [num_tokens, top_k]  (softmax over the chosen experts only)
          topk_idx:     [num_tokens, top_k]  (which experts were chosen)
          router_probs: [num_tokens, num_experts]  (full softmax, used for aux loss)
        """
        logits = self.gate(x_flat)                       # [tokens, num_experts]
        router_probs = F.softmax(logits, dim=-1)

        topk_weights, topk_idx = torch.topk(router_probs, self.top_k, dim=-1)
        # Re-normalize so the selected experts' weights sum to 1
        topk_weights = topk_weights / (topk_weights.sum(dim=-1, keepdim=True) + 1e-9)

        return topk_weights, topk_idx, router_probs


class MoEFeedForward(nn.Module):
    """Sparse Mixture-of-Experts feed-forward block: a drop-in replacement
    for a normal Transformer FFN."""

    def __init__(self, embed_dim: int, hidden_dim: int, num_experts: int,
                 top_k: int = 2):
        super().__init__()
        self.num_experts = num_experts
        self.top_k = top_k

        self.router = Router(embed_dim, num_experts, top_k)
        self.experts = nn.ModuleList(
            [Expert(embed_dim, hidden_dim) for _ in range(num_experts)]
        )

        # Populated on every forward pass; read by the training loop and
        # added to the main cross-entropy loss.
        self.last_aux_loss = torch.tensor(0.0)

        # Populated on every forward pass; how many tokens picked each expert
        # as their top-1 choice. Detached, used only for monitoring/eval
        # (expert-utilization / load-balance reporting), never for training.
        self.last_expert_counts = torch.zeros(num_experts)

    def _load_balancing_loss(self, router_probs: torch.Tensor, topk_idx: torch.Tensor) -> torch.Tensor:
        """Switch-Transformer style auxiliary loss: encourages the router to
        spread tokens evenly across experts instead of collapsing onto a few.

        For each expert e:
          f_e = fraction of tokens whose top-1 choice is e
          P_e = average router probability mass assigned to e (over all tokens)
        loss = num_experts * sum_e f_e * P_e
        This is minimized (== 1/num_experts... actually == num_experts * (1/N)^2 summed
        -> equals 1 when routing is perfectly uniform) when tokens are spread evenly.
        """
        num_tokens = router_probs.size(0)

        top1_idx = topk_idx[:, 0]
        tokens_per_expert = torch.zeros(self.num_experts, device=router_probs.device)
        tokens_per_expert.scatter_add_(
            0, top1_idx, torch.ones_like(top1_idx, dtype=router_probs.dtype)
        )
        f = tokens_per_expert / max(num_tokens, 1)

        P = router_probs.mean(dim=0)  # [num_experts]

        return self.num_experts * torch.sum(f * P)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        B, T, C = x.shape
        x_flat = x.view(-1, C)  # [B*T, C]

        topk_weights, topk_idx, router_probs = self.router(x_flat)

        output = torch.zeros_like(x_flat)

        # Dispatch tokens to each expert they were routed to. This loop is
        # over experts (not tokens), so the number of forward passes stays
        # fixed at num_experts regardless of batch size.
        for expert_id, expert in enumerate(self.experts):
            # positions (token_idx, slot_idx) where this expert was chosen
            token_idx, slot_idx = torch.where(topk_idx == expert_id)
            if token_idx.numel() == 0:
                continue

            expert_input = x_flat[token_idx]
            expert_output = expert(expert_input)

            weights = topk_weights[token_idx, slot_idx].unsqueeze(-1)
            output.index_add_(0, token_idx, expert_output * weights)

        self.last_aux_loss = self._load_balancing_loss(router_probs, topk_idx)

        with torch.no_grad():
            top1_idx = topk_idx[:, 0]
            counts = torch.zeros(self.num_experts, device=x.device)
            counts.scatter_add_(0, top1_idx, torch.ones_like(top1_idx, dtype=counts.dtype))
            self.last_expert_counts = counts.detach().cpu()

        return output.view(B, T, C)
````

````python
%%writefile ffn.py

"""
Dense (ordinary, non-MoE) feed-forward baseline.

Same SwiGLU shape as a single Expert in moe.py, so it's a drop-in for
`ffn_type="dense"` in model.py — used as the "no MoE" side of a MoE-vs-dense
comparison.
"""
import torch
import torch.nn as nn
import torch.nn.functional as F


class DenseFeedForward(nn.Module):
    """A single always-on SwiGLU FFN (no routing, no experts)."""

    def __init__(self, embed_dim: int, hidden_dim: int):
        super().__init__()
        self.gate_proj = nn.Linear(embed_dim, hidden_dim, bias=False)
        self.up_proj = nn.Linear(embed_dim, hidden_dim, bias=False)
        self.down_proj = nn.Linear(hidden_dim, embed_dim, bias=False)

        # Present so model.py can read .last_aux_loss / .last_expert_counts
        # uniformly for both FFN types without branching everywhere.
        self.last_aux_loss = torch.tensor(0.0)
        self.last_expert_counts = torch.zeros(1)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.down_proj(F.silu(self.gate_proj(x)) * self.up_proj(x))


def dense_hidden_dim_matching_active_compute(expert_hidden_dim: int, top_k: int) -> int:
    """For an iso-FLOPs-per-token comparison against a MoE FFN: a token
    passing through `top_k` experts of width `expert_hidden_dim` does the
    same amount of work as one dense FFN of width `top_k * expert_hidden_dim`.
    Use this so 'MoE vs dense' isn't secretly 'more compute vs less compute'.
    """
    return top_k * expert_hidden_dim
````

````python
%%writefile model.py

"""
Decoder-only transformer LLM, config-driven so the architecture can be
swapped for controlled comparisons:

  attention_type: "mha" -> standard Multi-Head Attention (every head has its
                            own K/V; implemented as GQA with num_kv_heads == num_heads)
                  "gqa" -> Causal Grouped Query Attention (attention.py)
                  "gla" -> Grouped Latent Attention (grouped_latent_attention.py) --
                            NOT Gated Linear Attention, see that module's docstring
  ffn_type:       "moe"   -> Mixture-of-Experts feed-forward (moe.py)
                  "dense" -> ordinary always-on MLP feed-forward (ffn.py)

Each block is: RMSNorm -> Attention (+residual) -> RMSNorm -> FFN (+residual).
The token-embedding matrix and the final output projection share the same
weight tensor ("weight tying").
"""
import torch
import torch.nn as nn

from attention import GroupedQueryAttention, RMSNorm
from grouped_latent_attention import GroupedLatentAttention
from moe import MoEFeedForward
from ffn import DenseFeedForward

FFN_ALIASES = {"mlp": "dense", "dense": "dense", "moe": "moe"}


def _build_attention(attention_type, embed_dim, num_heads, num_kv_heads,
                      num_latent_groups, latent_dim, max_seq_len, dropout):
    if attention_type == "mha":
        # Standard multi-head attention is GQA with one KV head per Q head
        return GroupedQueryAttention(embed_dim, num_heads, num_heads, max_seq_len, dropout)
    elif attention_type == "gqa":
        return GroupedQueryAttention(embed_dim, num_heads, num_kv_heads, max_seq_len, dropout)
    elif attention_type == "gla":
        return GroupedLatentAttention(embed_dim, num_heads, num_latent_groups, latent_dim, max_seq_len, dropout)
    raise ValueError(f"unknown attention_type: {attention_type!r} (expected 'mha', 'gqa', or 'gla')")


def _build_ffn(ffn_type, embed_dim, hidden_dim, num_experts, top_k):
    ffn_type = FFN_ALIASES.get(ffn_type, ffn_type)
    if ffn_type == "moe":
        return MoEFeedForward(embed_dim, hidden_dim, num_experts, top_k)
    elif ffn_type == "dense":
        return DenseFeedForward(embed_dim, hidden_dim)
    raise ValueError(f"unknown ffn_type: {ffn_type!r} (expected 'moe' or 'dense'/'mlp')")


class Block(nn.Module):
    def __init__(self, embed_dim, num_heads, num_kv_heads, num_latent_groups, latent_dim,
                 max_seq_len, num_experts, hidden_dim, top_k, dropout, attention_type, ffn_type):
        super().__init__()
        self.norm1 = RMSNorm(embed_dim)
        self.attn = _build_attention(attention_type, embed_dim, num_heads, num_kv_heads,
                                      num_latent_groups, latent_dim, max_seq_len, dropout)
        self.norm2 = RMSNorm(embed_dim)
        self.ffn = _build_ffn(ffn_type, embed_dim, hidden_dim, num_experts, top_k)

    def forward(self, x):
        x = x + self.attn(self.norm1(x))
        x = x + self.ffn(self.norm2(x))
        return x


class MoELLM(nn.Module):
    def __init__(
        self,
        vocab_size: int,
        embed_dim: int = 256,
        num_layers: int = 6,
        num_heads: int = 8,
        num_kv_heads: int = 4,          # only used when attention_type == "gqa"
        num_latent_groups: int = None,  # only used when attention_type == "gla"; default = num_kv_heads
        latent_dim: int = None,         # only used when attention_type == "gla"; default = 2 * head_dim
        max_seq_len: int = 128,
        num_experts: int = 8,           # only used when ffn_type == "moe"
        top_k: int = 2,                 # only used when ffn_type == "moe"
        hidden_dim: int = None,         # per-expert width (moe) or FFN width (dense)
        dropout: float = 0.0,
        attention_type: str = "gqa",    # "mha", "gqa", or "gla"
        ffn_type: str = "moe",          # "moe" or "dense"/"mlp"
    ):
        super().__init__()
        head_dim = embed_dim // num_heads
        num_latent_groups = num_latent_groups or num_kv_heads
        latent_dim = latent_dim or 2 * head_dim
        ffn_type = FFN_ALIASES.get(ffn_type, ffn_type)
        hidden_dim = hidden_dim or 4 * embed_dim

        self.max_seq_len = max_seq_len
        self.attention_type = attention_type
        self.ffn_type = ffn_type

        # Stash the exact kwargs used to build this model, so a checkpoint
        # can always be reloaded with MoELLM(**checkpoint["config"]).
        self.init_kwargs = dict(
            vocab_size=vocab_size, embed_dim=embed_dim, num_layers=num_layers,
            num_heads=num_heads, num_kv_heads=num_kv_heads,
            num_latent_groups=num_latent_groups, latent_dim=latent_dim,
            max_seq_len=max_seq_len, num_experts=num_experts, top_k=top_k,
            hidden_dim=hidden_dim, dropout=dropout,
            attention_type=attention_type, ffn_type=ffn_type,
        )

        self.token_embedding = nn.Embedding(vocab_size, embed_dim)
        self.dropout = nn.Dropout(dropout)

        self.blocks = nn.ModuleList([
            Block(embed_dim, num_heads, num_kv_heads, num_latent_groups, latent_dim,
                  max_seq_len, num_experts, hidden_dim, top_k, dropout, attention_type, ffn_type)
            for _ in range(num_layers)
        ])

        self.final_norm = RMSNorm(embed_dim)
        self.lm_head = nn.Linear(embed_dim, vocab_size, bias=False)

        # Weight tying: reuse the embedding matrix as the output projection
        self.lm_head.weight = self.token_embedding.weight

        self.apply(self._init_weights)

    @staticmethod
    def _init_weights(module):
        if isinstance(module, nn.Linear):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)
            if module.bias is not None:
                nn.init.zeros_(module.bias)
        elif isinstance(module, nn.Embedding):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)

    def forward(self, input_ids: torch.Tensor):
        """input_ids: [batch, seq_len] -> logits: [batch, seq_len, vocab_size]"""
        assert input_ids.size(1) <= self.max_seq_len, "sequence longer than max_seq_len"

        x = self.token_embedding(input_ids)
        x = self.dropout(x)

        for block in self.blocks:
            x = block(x)

        x = self.final_norm(x)
        logits = self.lm_head(x)

        # Only meaningful for ffn_type == "moe"; zero-valued (no-op) for "dense"
        aux_loss = torch.stack([b.ffn.last_aux_loss for b in self.blocks]).mean()

        return logits, aux_loss

    def expert_usage(self):
        """Per-block tensors of top-1 expert-choice counts. Only meaningful
        when ffn_type == 'moe'."""
        return [block.ffn.last_expert_counts for block in self.blocks]

    def num_active_params(self) -> int:
        """Parameters actually used per token (vs. total stored parameters)
        -- the fair number to compare against a dense model of matching
        per-token compute. Equals total params for ffn_type == 'dense'."""
        total = sum(p.numel() for p in self.parameters())
        if self.ffn_type != "moe":
            return total

        per_block_savings = 0
        for block in self.blocks:
            moe = block.ffn
            per_expert_params = sum(p.numel() for p in moe.experts[0].parameters())
            unused_experts = moe.num_experts - moe.top_k
            per_block_savings += per_expert_params * unused_experts
        return total - per_block_savings

    def kv_cache_elements_per_token(self) -> int:
        """Total KV-cache size (in scalar elements) needed per generated
        token, summed over all layers -- the actual quantity all three
        attention variants (MHA/GQA/GLA) are trying to shrink."""
        return sum(b.attn.kv_cache_elements_per_token() for b in self.blocks)

    @torch.no_grad()
    def generate(self, input_ids: torch.Tensor, max_new_tokens: int,
                 temperature: float = 0.8, top_k: int = 50, eos_token_id: int = None):
        """Autoregressive sampling. input_ids: [1, seq_len]"""
        self.eval()
        for _ in range(max_new_tokens):
            cond_ids = input_ids[:, -self.max_seq_len:]
            logits, _ = self(cond_ids)
            next_logits = logits[:, -1, :] / max(temperature, 1e-6)

            if top_k is not None:
                v, _ = torch.topk(next_logits, min(top_k, next_logits.size(-1)))
                next_logits[next_logits < v[:, [-1]]] = float("-inf")

            probs = torch.softmax(next_logits, dim=-1)
            next_token = torch.multinomial(probs, num_samples=1)

            input_ids = torch.cat([input_ids, next_token], dim=1)

            if eos_token_id is not None and next_token.item() == eos_token_id:
                break

        return input_ids
````

````python
%%writefile data.py

"""
Data pipeline for training the MoE LLM on WikiText-103.

Steps:
  1. Load raw WikiText-103 train/validation splits from Hugging Face
  2. Clean it (drop blank lines / "= Heading =" lines, strip <unk> markers,
     join articles with an <|endoftext|> separator so the model learns
     natural stopping points)
  3. Tokenize with GPT-2's BPE tokenizer (via tiktoken)
  4. Wrap the token stream in a PyTorch Dataset that yields
     (input_ids, target_ids) pairs shifted by one position

If the `datasets` library / Hugging Face Hub isn't reachable (e.g. no
internet access), `load_wikitext` falls back to a small bundled sample of
public-domain text so the rest of the pipeline can still be exercised
end-to-end. Swap in the real dataset for actual training.
"""
import os
import tiktoken
import torch
from torch.utils.data import Dataset

EOS_TOKEN = "<|endoftext|>"

_FALLBACK_TEXT = """
The mixture of experts approach to neural networks divides a large model
into many smaller sub-networks called experts . A learned router decides ,
for every token , which small number of experts should process it . Only
those experts run , so the model can have a very large total parameter
count while keeping the computation for each token cheap . This idea was
used in several modern large language models to scale up capacity without
a matching increase in inference cost .

Grouped query attention is a modification of multi head attention in which
several query heads share the same key and value projections . This
reduces the memory needed to cache keys and values during generation ,
which matters a great deal when serving long contexts . It sits between
multi head attention , where every head has its own keys and values , and
multi query attention , where all heads share a single key and value .
""".strip()


def clean_wikitext(dataset) -> str:
    """Drop blank lines and '= Heading =' lines, strip <unk> markers, and
    join everything with EOS separators."""
    cleaned = []
    for text in dataset["text"]:
        line = text.strip()
        if line and not line.startswith("="):
            cleaned.append(" ".join(line.replace("<unk>", "").split()))
    return EOS_TOKEN.join(cleaned) + EOS_TOKEN


def load_wikitext_split(split: str) -> str:
    """Loads and cleans a single WikiText-103 split ('train', 'validation',
    or 'test'). Falls back to a tiny bundled sample if the dataset can't be
    downloaded (e.g. no network access), split roughly so every split still
    gets some text for a quick offline smoke test."""
    try:
        from datasets import load_dataset

        raw = load_dataset("wikitext", "wikitext-103-v1", split=split)
        return clean_wikitext(raw)

    except Exception as err:
        print(f"[data] Falling back to bundled sample text for split='{split}' "
              f"({err.__class__.__name__}: {err})")
        text = _FALLBACK_TEXT + EOS_TOKEN
        n = len(text)
        bounds = {
            "train": (0, int(n * 0.7)),
            "validation": (int(n * 0.7), int(n * 0.85)),
            "test": (int(n * 0.85), n),
        }
        start, end = bounds.get(split, (0, n))
        return text[start:end]


def load_wikitext():
    """Convenience wrapper used by train.py: returns (training_text, validation_text)."""
    return load_wikitext_split("train"), load_wikitext_split("validation")


def get_tokenizer():
    return tiktoken.get_encoding("gpt2")


class TextDataset(Dataset):
    """Serves fixed-length (input_ids, target_ids) chunks of token ids,
    where target_ids is input_ids shifted forward by one position."""

    def __init__(self, text: str, tokenizer, max_seq_length: int):
        self.tokens = tokenizer.encode(text, allowed_special="all")
        self.max_seq_length = max_seq_length

    def __len__(self):
        return max(0, (len(self.tokens) - 1) // self.max_seq_length)

    def __getitem__(self, idx):
        start = idx * self.max_seq_length
        end = start + self.max_seq_length
        input_ids = torch.tensor(self.tokens[start:end], dtype=torch.long)
        target_ids = torch.tensor(self.tokens[start + 1:end + 1], dtype=torch.long)
        return input_ids, target_ids


````

````python
%%writefile train.py


"""
End-to-end training script for the LLM, config-driven so it can train any
combination of {mha, gqa, gla} attention x {moe, dense} FFN for controlled
comparisons.

Usage:
    python train.py                                        # defaults: gqa + moe
    python train.py --attention mha --ffn dense
    python train.py --attention gla --ffn moe --checkpoint runs/gla_moe.pt
"""
import argparse
import math

import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from torch.nn.utils import clip_grad_norm_
from torch.optim.lr_scheduler import CosineAnnealingLR
from tqdm import tqdm

from data import load_wikitext, get_tokenizer, TextDataset
from model import MoELLM
from ffn import dense_hidden_dim_matching_active_compute

# ----------------------------- Default config ------------------------------
MAX_SEQ_LENGTH = 128
BATCH_SIZE = 32
EMBED_DIM = 256
NUM_LAYERS = 6
NUM_HEADS = 8
NUM_KV_HEADS = 4          # GQA only: 8 query heads share 4 KV heads (group size 2)
NUM_LATENT_GROUPS = 4     # GLA only: like NUM_KV_HEADS but for latent groups
LATENT_DIM = None         # GLA only: default = 2 * head_dim (see model.py)
NUM_EXPERTS = 8           # MoE only
TOP_K = 2                 # MoE only: each token routed to its top-2 experts
EXPERT_HIDDEN_DIM = 4 * EMBED_DIM  # MoE only: width of each expert
AUX_LOSS_WEIGHT = 0.01    # MoE only: weight of the load-balancing loss
LEARNING_RATE = 3e-4
NUM_EPOCHS = 1
MAX_STEPS = None          # if set, stop after this many optimizer steps (overrides epochs)
GRAD_CLIP_NORM = 1.0
SEED = 0
# ---------------------------------------------------------------------------


def set_seed(seed):
    torch.manual_seed(seed)


def build_model(vocab_size, attention_type, ffn_type):
    ffn_type = "dense" if ffn_type == "mlp" else ffn_type

    if ffn_type == "moe":
        hidden_dim = EXPERT_HIDDEN_DIM
    else:
        # Iso-FLOPs-per-token: dense width = top_k * expert width, so a
        # MoE-vs-dense comparison isn't secretly a compute-budget difference.
        hidden_dim = dense_hidden_dim_matching_active_compute(EXPERT_HIDDEN_DIM, TOP_K)

    return MoELLM(
        vocab_size=vocab_size,
        embed_dim=EMBED_DIM,
        num_layers=NUM_LAYERS,
        num_heads=NUM_HEADS,
        num_kv_heads=NUM_KV_HEADS,
        num_latent_groups=NUM_LATENT_GROUPS,
        latent_dim=LATENT_DIM,
        max_seq_len=MAX_SEQ_LENGTH,
        num_experts=NUM_EXPERTS,
        top_k=TOP_K,
        hidden_dim=hidden_dim,
        attention_type=attention_type,
        ffn_type=ffn_type,
    )


def evaluate(model, val_loader, device):
    model.eval()
    total_loss, total_batches = 0.0, 0
    with torch.no_grad():
        for input_ids, target_ids in val_loader:
            input_ids, target_ids = input_ids.to(device), target_ids.to(device)
            logits, aux_loss = model(input_ids)
            loss = nn.functional.cross_entropy(
                logits.view(-1, logits.size(-1)), target_ids.view(-1)
            )
            total_loss += loss.item()
            total_batches += 1
    model.train()
    return total_loss / max(total_batches, 1)


def train_model(attention_type, ffn_type, epochs=1, max_steps=None, seed=SEED,
                 device=None, train_loader=None, val_loader=None, vocab_size=None,
                 verbose=True):
    """Trains one (attention_type, ffn_type) configuration and returns the
    trained model plus a small history dict. Shared by the CLI entrypoint
    below and by compare.py, so both use exactly the same training logic."""
    device = device or ("cuda" if torch.cuda.is_available() else "cpu")
    set_seed(seed)

    if train_loader is None or val_loader is None or vocab_size is None:
        training_text, validation_text = load_wikitext()
        tokenizer = get_tokenizer()
        vocab_size = tokenizer.n_vocab
        train_dataset = TextDataset(training_text, tokenizer, MAX_SEQ_LENGTH)
        val_dataset = TextDataset(validation_text, tokenizer, MAX_SEQ_LENGTH)
        train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
        val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False)

    model = build_model(vocab_size, attention_type, ffn_type).to(device)

    total_params = sum(p.numel() for p in model.parameters())
    active_params = model.num_active_params()
    if verbose:
        print(f"[train] {attention_type}+{ffn_type} | total params={total_params:,} "
              f"| active params/token={active_params:,}")

    optimizer = torch.optim.AdamW(model.parameters(), lr=LEARNING_RATE)
    total_steps = max_steps or max(len(train_loader) * epochs, 1)
    scheduler = CosineAnnealingLR(optimizer, T_max=total_steps)

    model.train()
    step = 0
    stop = False
    for epoch in range(epochs):
        if stop:
            break
        iterator = tqdm(train_loader, desc=f"{attention_type}+{ffn_type} epoch {epoch + 1}/{epochs}") if verbose else train_loader
        for input_ids, target_ids in iterator:
            input_ids, target_ids = input_ids.to(device), target_ids.to(device)

            logits, aux_loss = model(input_ids)
            ce_loss = nn.functional.cross_entropy(
                logits.view(-1, logits.size(-1)), target_ids.view(-1)
            )
            loss = ce_loss + (AUX_LOSS_WEIGHT * aux_loss if ffn_type == "moe" else 0.0)

            optimizer.zero_grad()
            loss.backward()
            clip_grad_norm_(model.parameters(), GRAD_CLIP_NORM)
            optimizer.step()
            scheduler.step()

            step += 1
            if max_steps is not None and step >= max_steps:
                stop = True
                break

        if not stop or epoch == epochs - 1:
            val_loss = evaluate(model, val_loader, device)
            if verbose:
                print(f"[train] {attention_type}+{ffn_type} epoch {epoch + 1} done "
                      f"| val_loss={val_loss:.4f} | val_ppl={math.exp(min(val_loss, 20)):.1f}")

    final_val_loss = evaluate(model, val_loader, device)
    history = {
        "steps_trained": step,
        "final_val_loss": final_val_loss,
        "final_val_ppl": math.exp(min(final_val_loss, 20)),
        "total_params": total_params,
        "active_params": active_params,
    }
    return model, history, vocab_size


def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--attention", choices=["mha", "gqa", "gla"], default="gqa")
    parser.add_argument("--ffn", choices=["moe", "dense", "mlp"], default="moe")
    parser.add_argument("--checkpoint", default=None,
                         help="Output path (default: checkpoint_<attention>_<ffn>.pt)")
    parser.add_argument("--epochs", type=int, default=NUM_EPOCHS)
    parser.add_argument("--max_steps", type=int, default=MAX_STEPS,
                         help="Stop after this many optimizer steps (for equal-compute comparisons)")
    parser.add_argument("--seed", type=int, default=SEED)
    args = parser.parse_args()

    checkpoint_path = args.checkpoint or f"checkpoint_{args.attention}_{args.ffn}.pt"

    print("[train] loading + cleaning WikiText ...")
    model, history, vocab_size = train_model(
        attention_type=args.attention,
        ffn_type=args.ffn,
        epochs=args.epochs,
        max_steps=args.max_steps,
        seed=args.seed,
    )

    torch.save(
        {
            "model_state_dict": model.state_dict(),
            "config": model.init_kwargs,
            "total_steps_trained": history["steps_trained"],
        },
        checkpoint_path,
    )
    print(f"[train] saved checkpoint -> {checkpoint_path}")


if __name__ == "__main__":
    main()
````

````python
%%writefile compare.py

"""
Train and compare multiple architecture configurations under controlled
conditions: same data, same seed, same number of steps.

Usage:
    python compare.py                                    # all 6: {mha,gqa,gla} x {moe,dense}
    python compare.py --configs gqa:moe gla:moe           # just these two
    python compare.py --max_steps 200                     # quick comparison run
    python compare.py --output_json comparison.json
"""
import argparse
import json

import torch
from torch.utils.data import DataLoader

from data import load_wikitext, get_tokenizer, TextDataset
from train import train_model, MAX_SEQ_LENGTH, BATCH_SIZE, SEED
from eval import measure_throughput

ALL_CONFIGS = ["mha:moe", "mha:dense", "gqa:moe", "gqa:dense", "gla:moe", "gla:dense"]


def parse_configs(config_strs):
    configs = []
    for c in config_strs:
        attention, ffn = c.split(":")
        ffn = "dense" if ffn == "mlp" else ffn
        assert attention in ("mha", "gqa", "gla"), f"unknown attention {attention!r}"
        assert ffn in ("moe", "dense"), f"unknown ffn {ffn!r}"
        configs.append((attention, ffn))
    return configs


def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--configs", nargs="+", default=ALL_CONFIGS,
                         help="e.g. --configs gqa:moe gla:dense")
    parser.add_argument("--epochs", type=int, default=1)
    parser.add_argument("--max_steps", type=int, default=None,
                         help="Cap every config to this many optimizer steps (recommended, "
                              "so the comparison is equal-compute rather than equal-epochs)")
    parser.add_argument("--seed", type=int, default=SEED)
    parser.add_argument("--save_checkpoints", action="store_true",
                         help="Save each trained model as checkpoint_<attn>_<ffn>.pt")
    parser.add_argument("--output_json", default="comparison.json")
    args = parser.parse_args()

    configs = parse_configs(args.configs)
    device = "cuda" if torch.cuda.is_available() else "cpu"
    print(f"[compare] device={device} | configs={args.configs}")

    # Build the data loaders ONCE and reuse them for every config, so every
    # model sees exactly the same batches in exactly the same order.
    print("[compare] loading + cleaning WikiText ...")
    training_text, validation_text = load_wikitext()
    tokenizer = get_tokenizer()
    vocab_size = tokenizer.n_vocab
    train_dataset = TextDataset(training_text, tokenizer, MAX_SEQ_LENGTH)
    val_dataset = TextDataset(validation_text, tokenizer, MAX_SEQ_LENGTH)
    print(f"[compare] {len(train_dataset):,} training sequences, "
          f"{len(val_dataset):,} validation sequences")

    results = []
    for attention_type, ffn_type in configs:
        name = f"{attention_type}+{ffn_type}"
        print(f"\n{'#' * 60}\n# Training {name}\n{'#' * 60}")

        # Fresh DataLoaders each time (same underlying dataset + seed) so
        # every config gets an identical shuffle order.
        torch.manual_seed(args.seed)
        train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
        val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False)

        model, history, _ = train_model(
            attention_type=attention_type,
            ffn_type=ffn_type,
            epochs=args.epochs,
            max_steps=args.max_steps,
            seed=args.seed,
            device=device,
            train_loader=train_loader,
            val_loader=val_loader,
            vocab_size=vocab_size,
        )

        throughput = measure_throughput(model, device)
        kv_cache = model.kv_cache_elements_per_token()

        result = {
            "config": name,
            "attention_type": attention_type,
            "ffn_type": ffn_type,
            "steps_trained": history["steps_trained"],
            "total_params": history["total_params"],
            "active_params": history["active_params"],
            "kv_cache_elements_per_token": kv_cache,
            "final_val_loss": history["final_val_loss"],
            "final_val_ppl": history["final_val_ppl"],
            "tokens_per_sec": throughput,
        }
        results.append(result)

        if args.save_checkpoints:
            path = f"checkpoint_{attention_type}_{ffn_type}.pt"
            torch.save({
                "model_state_dict": model.state_dict(),
                "config": model.init_kwargs,
            }, path)
            print(f"[compare] saved -> {path}")

        # free memory before training the next config
        del model
        if device == "cuda":
            torch.cuda.empty_cache()

    print_comparison_table(results)

    if args.output_json:
        with open(args.output_json, "w") as f:
            json.dump(results, f, indent=2)
        print(f"\n[compare] full results written -> {args.output_json}")


def print_comparison_table(results):
    print("\n" + "=" * 110)
    print("COMPARISON")
    print("=" * 110)
    header = (f"{'config':<14}{'steps':>8}{'total params':>15}{'active/tok':>13}"
              f"{'kv/tok':>10}{'val_loss':>10}{'val_ppl':>10}{'tok/sec':>12}")
    print(header)
    print("-" * len(header))
    for r in results:
        print(f"{r['config']:<14}{r['steps_trained']:>8}{r['total_params']:>15,}"
              f"{r['active_params']:>13,}{r['kv_cache_elements_per_token']:>10,}"
              f"{r['final_val_loss']:>10.4f}"
              f"{r['final_val_ppl']:>10.1f}{r['tokens_per_sec']:>12,.0f}")

    if len(results) > 1:
        best_loss = min(results, key=lambda r: r["final_val_loss"])
        best_speed = max(results, key=lambda r: r["tokens_per_sec"])
        smallest_cache = min(results, key=lambda r: r["kv_cache_elements_per_token"])
        print("-" * len(header))
        print(f"  lowest val_loss  : {best_loss['config']} ({best_loss['final_val_loss']:.4f})")
        print(f"  fastest          : {best_speed['config']} ({best_speed['tokens_per_sec']:,.0f} tok/s)")
        print(f"  smallest KV cache: {smallest_cache['config']} "
              f"({smallest_cache['kv_cache_elements_per_token']:,} elements/token)")


if __name__ == "__main__":
    main()
````

````python
%%writefile eval.py

"""
Evaluation harness for the MoE LLM.

Given a trained checkpoint, this reports:
  1. Cross-entropy loss, perplexity, and bits-per-token on a held-out split
     (test by default, falls back to validation if you prefer)
  2. Next-token top-1 accuracy
  3. Per-layer expert utilization: how evenly each MoE layer spreads tokens
     across its experts, summarized with a load-balance score (1.0 = perfect
     uniform usage, closer to num_experts = collapsed onto one expert)
  4. A handful of qualitative text-completion samples for a manual sanity check

Usage:
    python eval.py --checkpoint checkpoint.pt --split test
    python eval.py --checkpoint checkpoint.pt --split test --max_batches 50   # quick eval
"""
import argparse
import json
import math
import time

import torch
import torch.nn as nn
from torch.utils.data import DataLoader

from data import load_wikitext_split, get_tokenizer, TextDataset, EOS_TOKEN
from model import MoELLM

DEFAULT_PROMPTS = [
    "The history of",
    "In recent years, scientists have",
    "The mixture of experts approach",
]


def load_model(checkpoint_path: str, device: str) -> MoELLM:
    ckpt = torch.load(checkpoint_path, map_location=device)
    model = MoELLM(**ckpt["config"]).to(device)
    model.load_state_dict(ckpt["model_state_dict"])
    model.eval()
    return model


@torch.no_grad()
def measure_throughput(model, device, seq_len=None, batch_size=8, num_iters=10, warmup=2):
    """Rough tokens/sec for a forward pass at a given (batch_size, seq_len).
    Useful for comparing attention variants (e.g. GQA vs GLA) where the
    interesting difference is often speed, not just loss."""
    seq_len = seq_len or model.max_seq_len
    vocab_size = model.token_embedding.num_embeddings
    dummy = torch.randint(0, vocab_size, (batch_size, seq_len), device=device)

    for _ in range(warmup):
        model(dummy)
    if device == "cuda":
        torch.cuda.synchronize()

    start = time.perf_counter()
    for _ in range(num_iters):
        model(dummy)
    if device == "cuda":
        torch.cuda.synchronize()
    elapsed = time.perf_counter() - start

    tokens_processed = batch_size * seq_len * num_iters
    return tokens_processed / elapsed


@torch.no_grad()
def run_language_modeling_eval(model, loader, device, max_batches=None):
    """Computes loss / perplexity / bits-per-token / top-1 accuracy, and
    (for MoE models) accumulates per-layer expert usage counts, over a DataLoader."""
    total_loss, total_tokens, total_correct = 0.0, 0, 0
    is_moe = model.ffn_type == "moe"
    num_experts_per_layer = [b.ffn.num_experts for b in model.blocks] if is_moe else []
    usage_totals = [torch.zeros(n) for n in num_experts_per_layer]

    num_batches = 0
    for input_ids, target_ids in loader:
        if max_batches is not None and num_batches >= max_batches:
            break

        input_ids, target_ids = input_ids.to(device), target_ids.to(device)
        logits, _ = model(input_ids)

        flat_logits = logits.view(-1, logits.size(-1))
        flat_targets = target_ids.view(-1)

        loss = nn.functional.cross_entropy(flat_logits, flat_targets, reduction="sum")
        total_loss += loss.item()
        total_tokens += flat_targets.numel()

        preds = flat_logits.argmax(dim=-1)
        total_correct += (preds == flat_targets).sum().item()

        for i, usage in enumerate(model.expert_usage() if is_moe else []):
            usage_totals[i] += usage

        num_batches += 1

    avg_loss = total_loss / max(total_tokens, 1)
    perplexity = math.exp(min(avg_loss, 20))
    bits_per_token = avg_loss / math.log(2)
    accuracy = total_correct / max(total_tokens, 1)

    return {
        "batches_evaluated": num_batches,
        "tokens_evaluated": total_tokens,
        "loss": avg_loss,
        "perplexity": perplexity,
        "bits_per_token": bits_per_token,
        "top1_accuracy": accuracy,
        "expert_usage_counts": [u.tolist() for u in usage_totals],
    }


def load_balance_report(expert_usage_counts):
    """For each layer, turns raw usage counts into a human-readable balance
    summary. A perfectly balanced layer has every expert at 1/num_experts of
    the traffic; `balance_score` = num_experts * sum(p_e^2) is 1.0 when
    perfectly uniform and num_experts when totally collapsed onto one expert."""
    report = []
    for layer_idx, counts in enumerate(expert_usage_counts):
        counts = torch.tensor(counts)
        total = counts.sum().item()
        num_experts = counts.numel()

        if total == 0:
            fractions = torch.zeros(num_experts)
        else:
            fractions = counts / total

        balance_score = num_experts * (fractions.pow(2).sum().item())

        report.append({
            "layer": layer_idx,
            "usage_fractions": [round(f, 4) for f in fractions.tolist()],
            "balance_score": round(balance_score, 3),  # 1.0 = perfectly balanced
            "min_usage_pct": round(fractions.min().item() * 100, 2),
            "max_usage_pct": round(fractions.max().item() * 100, 2),
        })
    return report


@torch.no_grad()
def qualitative_samples(model, tokenizer, device, prompts, max_new_tokens=40):
    eos_id = tokenizer.encode(EOS_TOKEN, allowed_special="all")[0]
    samples = []
    for prompt in prompts:
        input_ids = torch.tensor(
            [tokenizer.encode(prompt, allowed_special="all")], dtype=torch.long
        ).to(device)
        output_ids = model.generate(
            input_ids, max_new_tokens=max_new_tokens,
            temperature=0.8, top_k=50, eos_token_id=eos_id,
        )
        samples.append({"prompt": prompt, "completion": tokenizer.decode(output_ids[0].tolist())})
    return samples


def print_report(metrics, balance_report, samples, arch_info=None):
    if arch_info:
        print("\n" + "=" * 60)
        print("ARCHITECTURE")
        print("=" * 60)
        print(f"  attention         : {arch_info['attention_type']}")
        print(f"  ffn               : {arch_info['ffn_type']}")
        print(f"  total params      : {arch_info['total_params']:,}")
        print(f"  active params/tok : {arch_info['active_params']:,}")
        print(f"  KV cache/token    : {arch_info['kv_cache_elements_per_token']:,} elements")
        print(f"  throughput        : {arch_info['tokens_per_sec']:,.0f} tokens/sec ({arch_info['device']})")

    print("\n" + "=" * 60)
    print("LANGUAGE MODELING METRICS")
    print("=" * 60)
    print(f"  batches evaluated : {metrics['batches_evaluated']}")
    print(f"  tokens evaluated  : {metrics['tokens_evaluated']:,}")
    print(f"  loss              : {metrics['loss']:.4f}")
    print(f"  perplexity        : {metrics['perplexity']:.2f}")
    print(f"  bits / token      : {metrics['bits_per_token']:.3f}")
    print(f"  top-1 accuracy    : {metrics['top1_accuracy']*100:.2f}%")

    print("\n" + "=" * 60)
    print("EXPERT LOAD-BALANCE PER LAYER  (1.0 = perfectly balanced)")
    print("=" * 60)
    for row in balance_report:
        bar = " ".join(f"{p*100:5.1f}%" for p in row["usage_fractions"])
        print(f"  layer {row['layer']:>2} | balance={row['balance_score']:.2f} "
              f"| min={row['min_usage_pct']:.1f}% max={row['max_usage_pct']:.1f}% "
              f"| usage: {bar}")

    print("\n" + "=" * 60)
    print("QUALITATIVE SAMPLES")
    print("=" * 60)
    for s in samples:
        print(f"  prompt:     {s['prompt']!r}")
        print(f"  completion: {s['completion']!r}\n")


def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--checkpoint", default="checkpoint.pt")
    parser.add_argument("--split", default="test", choices=["train", "validation", "test"])
    parser.add_argument("--batch_size", type=int, default=32)
    parser.add_argument("--max_batches", type=int, default=None,
                         help="Limit number of batches for a quick eval (default: full split)")
    parser.add_argument("--num_samples", type=int, default=3)
    parser.add_argument("--output_json", default=None,
                         help="Optional path to dump the full metrics/report as JSON")
    args = parser.parse_args()

    device = "cuda" if torch.cuda.is_available() else "cpu"
    print(f"[eval] device = {device}")

    model = load_model(args.checkpoint, device)
    tokenizer = get_tokenizer()

    print(f"[eval] loading '{args.split}' split ...")
    text = load_wikitext_split(args.split)
    dataset = TextDataset(text, tokenizer, model.max_seq_len)
    loader = DataLoader(dataset, batch_size=args.batch_size, shuffle=False)
    print(f"[eval] {len(dataset):,} evaluation sequences")

    metrics = run_language_modeling_eval(model, loader, device, args.max_batches)
    balance_report = load_balance_report(metrics.pop("expert_usage_counts"))
    samples = qualitative_samples(model, tokenizer, device, DEFAULT_PROMPTS[:args.num_samples])

    arch_info = {
        "attention_type": model.attention_type,
        "ffn_type": model.ffn_type,
        "total_params": sum(p.numel() for p in model.parameters()),
        "active_params": model.num_active_params(),
        "kv_cache_elements_per_token": model.kv_cache_elements_per_token(),
        "tokens_per_sec": measure_throughput(model, device),
        "device": device,
    }

    print_report(metrics, balance_report, samples, arch_info)

    if args.output_json:
        with open(args.output_json, "w") as f:
            json.dump({"architecture": arch_info, "metrics": metrics,
                       "expert_balance": balance_report, "samples": samples}, f, indent=2)
        print(f"[eval] full report written -> {args.output_json}")


if __name__ == "__main__":
    main()
````

````python
%%writefile generate.py

"""
Generate text from a trained MoE LLM checkpoint.

Usage:
    python generate.py --prompt "The mixture of experts" --tokens 60
"""
import argparse
import torch

from data import get_tokenizer, EOS_TOKEN
from model import MoELLM


def load_model(checkpoint_path: str, device: str):
    ckpt = torch.load(checkpoint_path, map_location=device)
    model = MoELLM(**ckpt["config"]).to(device)
    model.load_state_dict(ckpt["model_state_dict"])
    model.eval()
    return model


def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--checkpoint", default="checkpoint.pt")
    parser.add_argument("--prompt", default="The mixture of experts approach")
    parser.add_argument("--tokens", type=int, default=60)
    parser.add_argument("--temperature", type=float, default=0.8)
    parser.add_argument("--top_k", type=int, default=50)
    args = parser.parse_args()

    device = "cuda" if torch.cuda.is_available() else "cpu"
    tokenizer = get_tokenizer()
    eos_id = tokenizer.encode(EOS_TOKEN, allowed_special="all")[0]

    model = load_model(args.checkpoint, device)

    input_ids = torch.tensor(
        [tokenizer.encode(args.prompt, allowed_special="all")], dtype=torch.long
    ).to(device)

    output_ids = model.generate(
        input_ids,
        max_new_tokens=args.tokens,
        temperature=args.temperature,
        top_k=args.top_k,
        eos_token_id=eos_id,
    )

    print(tokenizer.decode(output_ids[0].tolist()))


if __name__ == "__main__":
    main()
````
